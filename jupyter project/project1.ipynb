```python
import numpy as np

class DataAnalytics:
    def __init__(self):
        self.array = None

    def create_array(self, dim, shape, elements):
        if dim == 1:
            self.array = np.array(elements)
        elif dim == 2:
            self.array = np.array(elements).reshape(shape[0], shape[1])
        elif dim == 3:
            self.array = np.array(elements).reshape(shape[0], shape[1], shape[2])
        print("\nArray created successfully:")
        print(self.array)

    def slice_array(self, row_start, row_end, col_start, col_end):
        if self.array is not None and self.array.ndim >= 2:
            sliced = self.array[row_start:row_end, col_start:col_end]
            print("\nSliced Array:")
            print(sliced)
        else:
            print("\nArray 2D up ")

    def math_operation(self, op_type, second_elements):
        second_array = np.array(second_elements).reshape(self.array.shape)
        print("\nOriginal Array:\n", self.array)
        print("Second Array:\n", second_array)

        if op_type == 1:
            res = self.array + second_array
            print("\nResult of Addition:\n", res)
        elif op_type == 2:
            res = self.array - second_array
            print("\nResult of Subtraction:\n", res)
        elif op_type == 3:
            res = self.array * second_array
            print("\nResult of Multiplication:\n", res)
        elif op_type == 4:
            res = self.array / second_array
            print("\nResult of Division:\n", res)

    def combine_array(self, second_elements):
        second_array = np.array(second_elements).reshape(1, self.array.shape[1])
        combined = np.vstack((self.array, second_array))
        print("\nOriginal Array:\n", self.array)
        print("Second Array:\n", second_array)
        print("\nCombined Array (Vertical Stack):\n", combined)


    def search_value(self, val):
        indices = np.where(self.array == val)
        print(f"\nValue {val} found at indices:", indices)

    def sort_array(self):
        sorted_arr = np.sort(self.array, axis=-1)
        print("\nOriginal Array:\n", self.array)
        print("\nSorted Array:\n", sorted_arr)

    def filter_array(self, threshold):
        filtered = self.array[self.array > threshold]
        print(f"\nValues greater than {threshold}:\n", filtered)


    def compute_stats(self, option):
        print("\nOriginal Array:\n", self.array)
        if option == 1:
            print("Sum of Array:", np.sum(self.array))
        elif option == 2:
            print("Mean of Array:", np.mean(self.array))
        elif option == 3:
            print("Median of Array:", np.median(self.array))
        elif option == 4:
            print("Standard Deviation:", np.std(self.array))
        elif option == 5:
            print("Variance:", np.var(self.array))



def main():
    analyzer = DataAnalytics()

    while True:
        print("\n=================================")
        print("Welcome to the NumPy Analyzer!")
        print("=================================")
        print("1. Create a Numpy Array")
        print("2. Perform Mathematical Operations")
        print("3. Combine or Split Arrays")
        print("4. Search, Sort, or Filter Arrays")
        print("5. Compute Aggregates and Statistics")
        print("6. Exit")
        
        choice = input("Enter your choice: ")

        if choice == '1':
            print("\nSelect the type of array to create:")
            print("1. 1D Array\n2. 2D Array\n3. 3D Array")
            dim = int(input("Enter your choice: "))
            
            if dim == 1:
                elements = list(map(int, input("Enter elements separated by space: ").split()))
                analyzer.create_array(1, None, elements)
            elif dim == 2:
                r = int(input("Enter the number of rows: "))
                c = int(input("Enter the number of columns: "))
                elements = list(map(int, input(f"Enter {r*c} elements separated by space: ").split()))
                analyzer.create_array(2, (r, c), elements)
            elif dim == 3:
                d1 = int(input("Enter dim 1: "))
                d2 = int(input("Enter dim 2: "))
                d3 = int(input("Enter dim 3: "))
                elements = list(map(int, input(f"Enter {d1*d2*d3} elements separated by space: ").split()))
                analyzer.create_array(3, (d1, d2, d3), elements)

        elif choice == '2':
            if analyzer.array is None:
                print("\nPehle Array create karo!")
                continue
            print("\nChoose a mathematical operation:")
            print("1. Addition\n2. Subtraction\n3. Multiplication\n4. Division")
            op = int(input("Enter your choice: "))
            size = analyzer.array.size
            elements = list(map(int, input(f"Enter same-size array elements ({size} elements): ").split()))
            analyzer.math_operation(op, elements)

        elif choice == '3':
            if analyzer.array is None:
                print("\nPehle Array create karo!")
                continue
            print("\nChoose an option:")
            print("1. Combine Arrays")
            ch = int(input("Enter your choice: "))
            if ch == 1:
                cols = analyzer.array.shape[1]
                elements = list(map(int, input(f"Enter {cols} elements to combine: ").split()))
                analyzer.combine_array(elements)

        elif choice == '4':
            if analyzer.array is None:
                print("\nPehle Array create karo!")
                continue
            print("\nChoose an option:")
            print("1. Search a value\n2. Sort the array\n3. Filter values")
            ch = int(input("Enter your choice: "))
            if ch == 1:
                val = int(input("Enter value to search: "))
                analyzer.search_value(val)
            elif ch == 2:
                analyzer.sort_array()
            elif ch == 3:
                th = int(input("Enter threshold value for filter (> threshold): "))
                analyzer.filter_array(th)

        elif choice == '5':
            if analyzer.array is None:
                print("\nPehle Array create karo!")
                continue
            print("\nChoose an aggregate/statistical operation:")
            print("1. Sum\n2. Mean\n3. Median\n4. Standard Deviation\n5. Variance")
            ch = int(input("Enter your choice: "))
            analyzer.compute_stats(ch)

        elif choice == '6':
            print("\nThank you for using the NumPy Analyzer! Goodbye!")
            break
        else:
            print("\nInvalid choice! Please try again.")

if __name__ == "__main__":
    main()
```
