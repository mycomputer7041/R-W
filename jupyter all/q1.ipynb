```python
import numpy as np
import pandas as pd

age=np.random.randint(1,100,size=101)
salary=np.random.randint(5000,60000,size=101)
des=np.random.choice(['ml eng','data anl','data sic'],size=101)
df=pd.DataFrame({
    'age':age,
    'salary':salary,
    'des':des
})

print(df)
print(df.head(10))

table=df.pivot_table(columns='des',values='salary',aggfunc='min')
print(table)

```
