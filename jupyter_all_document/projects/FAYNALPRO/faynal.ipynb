```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

df = pd.read_csv('train.csv')

print("\nDataset Summary")
print(df.info())

print("First 5 Row")
print(df.head())

print("\nMissing Values Count")
print(df.isnull().sum())





df['Age'] = df['Age'].fillna(df['Age'].median())


df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])


df = df.drop(columns=['Cabin'])


sns.set_theme(style='whitegrid')

def show_overall_survival():
    print("who can peapal can servay")
    plt.figure(figsize=(6, 4))
    sns.countplot(x='Survived', data=df, palette='Set2')
    plt.title('Overall Survival Count (0 = Died, 1 = Survived)')
    plt.xlabel('Survived')
    plt.ylabel('Passenger Count')
    plt.show()


def show_gender_survival():
    print("people servay by gender")
    plt.figure(figsize=(6, 4))
    sns.countplot(x='Sex', hue='Survived', data=df, palette='Set1')
    plt.title('Survival Rate by Gender')
    plt.xlabel('Gender')
    plt.ylabel('Count')
    plt.show()

def show_class_survival():
    print("people servay by class types")
    plt.figure(figsize=(6, 4))
    sns.countplot(x='Pclass', hue='Survived', data=df, palette='Set3')
    plt.title('Survival Rate by Passenger Class')
    plt.xlabel('Passenger Class (1st, 2nd, 3rd)')
    plt.ylabel('Count')
    plt.show()

def show_age_distribution():    
    print("people servay by age")
    plt.figure(figsize=(8, 4))
    sns.histplot(df['Age'], kde=True, bins=30, color='blue')
    plt.title('Age Distribution of Passengers')
    plt.xlabel('Age')
    plt.ylabel('Count')
    plt.show()


while True:
  print('\nTITANIC DATA VISUALIZATION MENU')
  print('1. Overall Survival Count ')
  print('2. Survival Rate by Gender ')
  print('3. Survival Rate by Passenger class')
  print('4. Age Distribution ')
  print('0. Exit ')

  choice = input('\nenter your choice in (0-4): ')

  if choice == '1':
    show_overall_survival()
  elif choice == '2':
    show_gender_survival()
  elif choice == '3':
    show_class_survival()
  elif choice == '4':
    show_age_distribution()
  elif choice == '0':
    print('Program is over')
    break
  else:
    print("enter valid number")
```
