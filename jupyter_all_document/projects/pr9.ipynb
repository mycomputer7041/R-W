```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

class SalesDataAnalyzer:
    def __init__(self):
        self.data = None  

    
    def load_dataset(self):
        path = input("Enter the path of the dataset  (sales_data.csv): ")
        try:
            self.data = pd.read_csv(path)
            print("\nDataset loaded successfully!")
        except Exception as e:
            print("\nError in loading file:", e)

    
    def explore_data(self):
        if self.data is None:
            print("\nfirs enter data set")
            return
        
        while True:
            print("\n== Explore Data Menu ==")
            print("1. Display first 5 rows")
            print("2. Display last 5 rows")
            print("3. Display column names")
            print("4. Display data types")
            print("5. Display basic info")
            print("6. Back to Main Menu")
            
            choice = input("Enter your choice (1-6): ")
            
            if choice == '1':
                print("\n", self.data.head())
            elif choice == '2':
                print("\n", self.data.tail())
            elif choice == '3':
                print("\nColumns:", self.data.columns.tolist())
            elif choice == '4':
                print("\n", self.data.dtypes)
            elif choice == '5':
                print("\n")
                self.data.info()
            elif choice == '6':
                break
            else:
                print("Invalid choice retry .")


    def perform_dataframe_operations(self):
        if self.data is None:
            print("\nfirst enter data set")
            return
        
        print("\n== DataFrame Operations ==")
        if 'Sales' in self.data.columns:
            total_sales = self.data['Sales'].sum()
            mean_sales = self.data['Sales'].mean()
            print(f"Total Sales (Sum): {total_sales}")
            print(f"Average Sales (Mean): {mean_sales}")
        else:
            print("Dataset in 'Sales' column can not find.")

 
    def handle_missing_data(self):
        if self.data is None:
            print("\nenetr first data set")
            return
        
        while True:
            print("\n== Handle Missing Data Menu ==")
            print("1. Check missing values count")
            print("2. Fill missing values with mean")
            print("3. Drop rows with missing values")
            print("4. Back to Main Menu")
            
            choice = input("Enter your choice (1-4): ")
            
            if choice == '1':
                print("\nMissing values in each column:\n", self.data.isnull().sum())
            elif choice == '2':
                self.data.fillna(self.data.mean(numeric_only=True), inplace=True)
                print("\nMissing values filled with mean successfully!")
            elif choice == '3':
                self.data.dropna(inplace=True)
                print("\nRows with missing values dropped successfully!")
            elif choice == '4':
                break
            else:
                print("Invalid choice.")

   
    def generate_statistics(self):
        if self.data is None:
            print("\nfirst enter data set")
            return
        print("\n== Descriptive Statistics ==")
        print(self.data.describe())

   
    def data_visualization(self):
        if self.data is None:
            print("\n first enter data set")
            return
        
        print("\n== Data Visualization Menu ==")
        print("1. Histogram (Sales distribution)")
        print("2. Scatter Plot")
        choice = input("Enter your choice (1-2): ")
        
        if choice == '1':
            plt.figure(figsize=(6, 4))
            plt.hist(self.data['Sales'].dropna(), bins=5, color='skyblue', edgecolor='black')
            plt.title('Sales Histogram')
            plt.xlabel('Sales')
            plt.ylabel('Frequency')
            plt.show()
            print("Histogram displayed successfully!")
        elif choice == '2':
            x_col = input("Enter X-axis column name (e.g., SalesID): ")
            y_col = input("Enter Y-axis column name (e.g., Sales): ")
            if x_col in self.data.columns and y_col in self.data.columns:
                plt.figure(figsize=(6, 4))
                plt.scatter(self.data[x_col], self.data[y_col], color='red')
                plt.title(f'Scatter Plot: {x_col} vs {y_col}')
                plt.xlabel(x_col)
                plt.ylabel(y_col)
                plt.show()
                print("Scatter plot displayed successfully!")
            else:
                print("Column names are not proper.")

    
    def save_visualization(self):
        filename = input("Enter file name to save plot (e.g.plot.png): ")
        plt.savefig(filename)
        print(f"Visualization saved as {filename} successfully!")
```

```python
if __name__ == "__main__":
    analyzer = SalesDataAnalyzer()
    
    while True:
        print("\n========== Data Analysis & Visualization Program ==========")
        print("Please select an option:")
        print("1. Load Dataset")
        print("2. Explore Data")
        print("3. Perform DataFrame Operations")
        print("4. Handle Missing Data")
        print("5. Generate Descriptive Statistics")
        print("6. Data Visualization")
        print("7. Save Visualization")
        print("8. Exit")
        print("==========================================================")
        
        choice = input("Enter your choice (1-8): ")
        
        if choice == '1':
            analyzer.load_dataset()
        elif choice == '2':
            analyzer.explore_data()
        elif choice == '3':
            analyzer.perform_dataframe_operations()
        elif choice == '4':
            analyzer.handle_missing_data()
        elif choice == '5':
            analyzer.generate_statistics()
        elif choice == '6':
            analyzer.data_visualization()
        elif choice == '7':
            analyzer.save_visualization()
        elif choice == '8':
            print("Exiting the program")
            break
        else:
            print("Invalid choice! Please enter a number between 1 and 8.")
```
