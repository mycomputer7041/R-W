```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

class FitnessTracker: #main class
    def __init__(self, filename="fitness_activities.csv"):
        self.filename = filename
        self.load_data()
    def load_data(self):# csv file & manyuly input
        try:
            self.df = pd.read_csv(self.filename) # csv file
        except FileNotFoundError:
            self.df = pd.DataFrame(columns=["Date", "Activity Type", "Duration", "Calories Burned"])#manyuly entry

    def log_activity(self, date, activity_type, duration, calories):
        if duration <= 0 or calories <= 0:
            print("duration and calories are postiv is very important so you writing the positiv number!")
            return
        self.df.loc[len(self.df)] = [date, activity_type, duration, calories]
        self.df.to_csv(self.filename, index=False)
        print("Activity successfully save")

    def calculate_metrics(self):
        total_calories = self.df["Calories Burned"].sum()
        avg_duration = self.df["Duration"].mean()
        print(f"Total Calories Burned: {total_calories}")
        print(f"Average Duration: {avg_duration:.2f} minutes")

    def filter_activities(self, activity_type):
        result = self.df[self.df["Activity Type"] == activity_type]
        return result

    def generate_report(self):
        print("\n Fitness Report")
        print(self.df.describe())

    def visualize_data(self):
        plt.figure(figsize=(12, 10))

    def show_bar_chart(self):
        plt.figure(figsize=(6, 4))
        total_time = self.df.groupby("Activity Type")["Duration"].sum()
        total_time.plot(kind="bar", color="skyblue")
        plt.title("Time Spent on Each Activity")
        plt.xlabel("Activity Type")
        plt.ylabel("Total Duration")
        plt.show()


    def show_line_graph(self):
        plt.figure(figsize=(6, 4))
        dates = pd.to_datetime(self.df["Date"])
        calories = self.df["Calories Burned"]
        plt.plot(dates, calories, marker="o", color="green")
        plt.title("Calories Burned Over Time")
        plt.xlabel("Date")
        plt.ylabel("Calories")
        plt.show()

    
    def show_pie_chart(self):
        plt.figure(figsize=(6, 4))
        activity_counts = self.df["Activity Type"].value_counts()
        activity_counts.plot(kind="pie", autopct="%1.1f%%", startangle=90)
        plt.title("Activity Distribution")
        plt.ylabel("")
        plt.show()
```

```python
if __name__ == "__main__":
    tracker = FitnessTracker()
    
    while True:
        print("\nFitness Tracker Dashboard Menu")
        print("1. Report View ")
        print("2. Nawi Activity Log ")
        print("3. Metrics Calculate ")
        print("4. Activity Filter ")
        print("5. Bar Chart ")
        print("6. Line Graph ")
        print("7. Pie Chart ")
        print("8. Exit")
       
        
        choice = input("choice 1 to 8: ")
        
        if choice == '1':
            tracker.generate_report()
        elif choice == '2':
            date = input("enter the your date(YYYY-MM-DD): ")
            activity_type = input("Activity Type (Running, Yoga): ")
        
            duration = float(input("Duration (minutes): "))
            calories = float(input("Calories Burned: "))
            tracker.log_activity(date, activity_type, duration, calories)
            
        elif choice == '3':
            tracker.calculate_metrics()
        elif choice == '4':
            act_type = input("who activity filter you: ")
            print(tracker.filter_activities(act_type))
        elif choice == '5':
            tracker.show_bar_chart()
        elif choice == '6':
            tracker.show_line_graph()
        elif choice == '7':
            tracker.show_pie_chart()
        elif choice == '8':
            break
        else:
            print("this option not valid")
```
