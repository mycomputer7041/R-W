```python
import matplotlib.pyplot as plt

#bar chart

plt.figure(figsize=(6,7))

subject=["maths","gujrati","english"]
marks=[30,40,35]

plt.bar(subject,marks,width=0.5,color=['red','green','blue'],alpha=0.8,align="center",edgecolor="black",linewidth=2)
plt.title("subject marks")
plt.xlabel("subject")
plt.ylabel("marks")

plt.text(0,30+1,str(30))
plt.text(1,40+1,str(40))
plt.text(2,35+1,str(35))

plt.show()
```

```python
#line vhart
plt.figure(figsize=(5,5))
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
          'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
revenue = [5000, 5200, 5800, 6200, 6000, 6700, 7100, 6900, 7200, 7500, 7700, 8000]

plt.plot(months,revenue,color="red",marker="o",markersize=5,markerfacecolor="green")
plt.xlabel("months")
plt.ylabel("revenue")
plt.title("months and revenue gunrets")
```
